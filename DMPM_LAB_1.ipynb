```python
def mean(nums:list[int])->int:
  sum=0
  for el in nums:
    sum+=el
  return sum//len(nums)
```

```python
def mode(nums:list[int])->int:
  freq_map={}#create a dict
  for el in nums:#run a loop on each element
    freq_map[el] = freq_map.get(el, 0) + 1#store alll the elemnt with their frequency

  max_count=0
  mode_val=None#create 2 variable

  for num, count in freq_map.items():#run a loop on key and value
    if count>max_count:#compare a with each elment and store the value
      max_count=count
      mode_val=num

  return mode_val#at the end return the value
```

```python
def median(nums:list[int])->int:
  nums.sort()#sort the array
  n=len(nums)
  if n%2==0:# if odd then return this
    pos=(n+1)//2
    return nums[pos]
  else:#else this
    return nums[n//2]
```

```python
def variance(nums:list[int])->int:
  m=mean(nums)#compute the mean
  sum=0#create a sum var
  for el in nums:
    sum+=(el-m)**2#compute and store in sum
  return sum/len(nums)#return the sum/lengt
```

```python
def standarddeviation(nums:list[int])->int:
  return variance(nums)**0.5#compute var and rooot it
```

```python
import numpy as np

nums = np.random.randint(0, 100, size=10)
print(nums)
```

```python
print('Mean=',mean(nums))
print('Median=',median(nums))
print('Mode',mode(nums))
print('Variance=',variance(nums))
print('Standard Deviation=',standarddeviation(nums))
```
