```python

numeric_df = df.select_dtypes(include=['number'])
corr = numeric_df.corr()
plt.figure(figsize=(10, 8))
sns.heatmap(corr, annot=True)
```

```python

plt.figure(figsize=(10, 8))

vis=sns.countplot(x = df['Status'] , data = df , palette = ['skyblue','r'])


plt.title('Target Distribution')
plt.show()
```

```python
#scatterplot
plt.figure(figsize=(10, 6))
sns.scatterplot(x='Age', y='Survival Months', hue='Status', data=df, alpha=0.7)
plt.title('Age vs. Survival Months by Status')
plt.xlabel('Age')
plt.ylabel('Survival Months')
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```

```python
plt.figure(figsize=(8, 6))
sns.boxplot(x='Status', y='Survival Months', data=df, palette={'Alive': 'g', 'Dead': 'r'})
plt.title('Survival Months Distribution by Status')
plt.xlabel('Status')
plt.ylabel('Survival Months')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

```python
#histogram
plt.figure(figsize=(10, 6))
sns.histplot(df['Age'], bins=20,kde=True, color='skyblue')
plt.title('Age Distribution')
plt.xlabel('Age')
plt.ylabel('Frequency')
```

```python
#violin plot
plt.figure(figsize=(10, 6))
sns.violinplot(x='Status', y='Age', data=df, palette={'Alive': 'skyblue', 'Dead': 'r'})
plt.title('Age Distribution by Status')
plt.xlabel('Status')
plt.ylabel('Age')
plt.grid(axis='y', linestyle='--', alpha=0.7)
```
