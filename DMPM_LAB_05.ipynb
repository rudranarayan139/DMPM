# Appriori Algorithm

> The Apriori algorithm is a classical unsupervised machine learning and data mining technique used for association rule learning



```python
pip install pandas mlxtend
```

```python
import pandas as pd
from mlxtend.frequent_patterns import apriori, association_rules
from mlxtend.preprocessing import TransactionEncoder
dataset = [
    ['Milk', 'Bread', 'Butter'],
    ['Bread', 'Butter', 'Diaper'],
    ['Milk', 'Diaper', 'Beer', 'Cola'],
    ['Milk', 'Bread', 'Butter', 'Diaper'],
    ['Bread', 'Milk', 'Diaper']
]
te = TransactionEncoder()
te_ary = te.fit(dataset).transform(dataset)
df = pd.DataFrame(te_ary, columns=te.columns_)
print("One-Hot Encoded Transaction Table")
print(df)
print("\n" + "="*50 + "\n")
frequent_itemsets = apriori(df, min_support=0.6, use_colnames=True)
print("--- Frequent Itemsets ---")
print(frequent_itemsets)
print("\n" + "="*50 + "\n")
rules = association_rules(frequent_itemsets, metric="confidence", min_threshold=0.7)
rules_summary = rules[['antecedents', 'consequents', 'support', 'confidence', 'lift']]
print("--- Association Rules Generated ---")
print(rules_summary.to_string())
```

# FP-Growth (Frequent Pattern Growth) algorithm
it's a highly efficient data mining method used to find frequent itemsets in transactional databases without the costly "candidate generation" step required by Apriori

```python
import pandas as pd
from mlxtend.frequent_patterns import fpgrowth
dataset = [['Milk', 'Bread', 'Butter'],
           ['Milk', 'Bread'],
           ['Milk', 'Diapers', 'Beer'],
           ['Milk', 'Bread', 'Butter', 'Diapers', 'Beer'],
           ['Bread', 'Butter', 'Diapers']]

from mlxtend.preprocessing import TransactionEncoder
te = TransactionEncoder()
te_ary = te.fit(dataset).transform(dataset)
df = pd.DataFrame(te_ary, columns=te.columns_)

frequent_itemsets = fpgrowth(df, min_support=0.4, use_colnames=True)

print(frequent_itemsets.sort_values(by="support", ascending=False))
```
