```python
import seaborn as sns
from sklearn.datasets import make_blobs
X,y=make_blobs(
    n_samples=1000,
    n_features=2,
    centers=4,
    random_state=42
)
```

```python

sns.scatterplot(x=X[:,0],y=X[:,1])
```

```python
from sklearn.cluster import KMeans
K=4
kmeans=KMeans(
    n_clusters=K,
    random_state=42

)
pred=kmeans.fit_predict(X)
```

```python
sns.scatterplot(x=X[:,0],y=X[:,1],c=pred)
```

```python
# Finding K value by Elbow Inertia Method
wss=[]
for k in range(1,21):
  kmean=KMeans(n_clusters=k,random_state=42)
  kmean.fit_predict(X)
  wss.append(kmean.inertia_)
sns.scatterplot(x=range(1,21),y=wss,marker="o")
```

```python
#Using Silhoute tech
from sklearn.metrics import silhouette_score
ss=[]
for k in range(2,21):
  model=KMeans(n_clusters=k)
  labels=model.fit_predict(X)
  ss=silhouette_score(X,labels)
sns.scatterplot(x=range(2,21),y=ss,markers="o")
```
