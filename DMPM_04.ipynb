Min-Max Normalization

```python
import numpy as np
x=np.array([10,20,30,40,50])
```

```python
import numpy as np

mn = np.min(x)
mx = np.max(x)
z = 25
st_range = 0
end_range = 1

if (mx - mn) == 0:
    normalized_value = st_range
else:
    res = (z - mn) / (mx - mn)
    normalized_value = (res * (end_range - st_range)) + st_range

print(f"Original array x: {x}")
print(f"Min of x: {mn}")
print(f"Max of x: {mx}")
print(f"Element to normalize (z): {z}")
print(f"Target start range: {st_range}")
print(f"Target end range: {end_range}")
print(f"Normalized value for {z}: {normalized_value}")
```

Z-Score

```python
def z_score(data):
  return (data - np.mean(data)) / np.std(data)
```

Decimal-Scaling

```python
def decimal_scaling(data):
  data = np.asarray(data)
  max_abs_val = np.max(np.abs(data))
  if max_abs_val == 0:
    return data
  divisor_power = len(str(int(max_abs_val)))
  divisor = 10**divisor_power
  return data / divisor
```

```python
print('Z_Score=',z_score(x))
print('Decimal_scalling=',decimal_scaling(x))
```

chi-Squre Test

```python
from scipy import stats
import numpy as np
def chi_squre(data):
  chi2, p, dof, expected = stats.chi2_contingency(data)
  return chi2, p, dof, expected
```
